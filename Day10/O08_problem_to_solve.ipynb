```python
import numpy as np
import pandas as pd
```

```python
r_cols = ['user_id', 'movie_id', 'ratings']
ratings = pd.read_csv("./assets/u.data", sep="\t", usecols=range(3), names=r_cols, encoding='ISO-8859-1')
print(ratings.head())
m_cols = ['movie_id', 'title']
movies = pd.read_csv("./assets/u.item", sep="|", usecols=range(2), names=m_cols, encoding='ISO-8859-1')
print(movies.head())
print("Shape of movies: ")
print(movies.shape)
print("="*60)
ratings = pd.merge(movies, ratings)
print(ratings.head())
print(ratings.shape)
print("=" * 60)
```

```python
print("="*60)
moviesRatings = ratings.pivot_table(index=['user_id'], columns=['title'], values='ratings')
print(moviesRatings.head())
```

```python
myMovies = moviesRatings['Star Wars (1977)']
print(myMovies.head())
```

```python
similarMovies = moviesRatings.corrwith(myMovies)
similarMovies = similarMovies.dropna()
df = pd.DataFrame(similarMovies)
print(df.head())
```

```python

print(similarMovies.sort_values(ascending=False))
```

```python
import numpy as np
movieStats = ratings.groupby('title').agg({'ratings': [np.size, np.mean]})
print(movieStats.head())
```

```python
popularMovies = movieStats['ratings']['size'] >= 100
print(movieStats[popularMovies].sort_values([('ratings', 'mean')], ascending=False)[:15])
```
