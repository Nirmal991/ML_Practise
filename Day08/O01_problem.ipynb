```python
import pandas as pd
```

```python
data = {
    "temperature":[-10, -5, 0, 5, 10, 15, 20, 25, 30, 35],
    'Cost': [220, 160, 110, 70, 45, 35, 40, 65, 105, 160]
}
df = pd.DataFrame(data)
print(df.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
plt.plot(df['temperature'], df['Cost'])
plt.xlabel('Temprature')
plt.ylabel('Cost')
plt.title('Weather Expense')
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
X = df[['temperature']]
y = df['Cost']

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)
print()
```

```python
new_data_to_predict ={
    "temperature":[-10, -5, 0, -15, -7, 3, 12, 18, 28, 33]
}
new_df = pd.DataFrame(new_data_to_predict)
```

```python
X_new = new_df[['temperature']]
X_new_poly = poly.transform(X_new)
predictions = model.predict(X_new_poly)
# print(predictions)
new_df['Predicted_Cost'] = predictions
print(new_df.to_string(index=False))
```

```python
merged_df = df.copy()
merged_df['Predicted_Cost'] = predictions
print(merged_df.to_string(index=False))
```
