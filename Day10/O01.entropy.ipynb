```python
import math
from collections import Counter
```

```python
'''
formula for entropy is : 
H(X) = -∑ P(x) log2 P(x)
'''
def calculate_entropy(labesl):
    total_count = len(labesl)
    counts = Counter(labesl)
    entropy = 0.0
    for count in counts.values():
        probability = count / total_count
        entropy -= probability * math.log2(probability)
    return entropy
```

```python
pure_group = ["Iguana", "Iguana", "Iguana", "Iguana"]
mixed_group = ["Iguana", "Iguana", "Dog", "Dog"]
print(f"Entropy of mixed groups is: {calculate_entropy(mixed_group)}")
print(f"Entropy of pure groups is: {calculate_entropy(pure_group)}")
```

```python
from scipy.stats import entropy
from collections import Counter

h_mixed = entropy([0.5,0.5],base=2)
print(f"Entropy of mixed groups is: {h_mixed}")
h_pure = entropy([1.0,0.0],base=2)
print(f"Entropy of pure groups is: {h_pure}")
```

```python
pure_group = ['Iguana', 'Iguana', 'Iguana', 'Iguana']
pure_counts = list(Counter(pure_group).values())  
print("Pure Group Entropy:", entropy(pure_counts, base=2))

# --- Case 2: High Entropy (50/50 split) ---
mixed_group = ['Iguana', 'Iguana', 'Dog', 'Dog']
mixed_counts = list(Counter(mixed_group).values())  # [2, 2]
print("50/50 Mixed Group Entropy:", entropy(mixed_counts, base=2))

# --- Case 3: Low Entropy (9:1 split) ---
mixed_group_low = ['Iguana'] * 9 + ['Dog']
mixed_counts_low = list(Counter(mixed_group_low).values())  # [9, 1]
print("9:1 Mixed Group Entropy:", entropy(mixed_counts_low, base=2))
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier,export_text

df = pd.DataFrame([
    {"wickets": 5, "role": "Batsman"},
    {"wickets": 8, "role": "Batsman"},
    {"wickets": 337, "role": "Bowler"},
    {"wickets": 501, "role": "Bowler"},
])

X = df[["wickets"]]
y = df["role"]

clf = DecisionTreeClassifier(criterion="entropy", random_state=42)
# print("CLF: ", clf)
clf.fit(X,y)

print("predictions: ", clf.predict(pd.DataFrame({"wickets": [10,400]})))
print("="*60)
print("Disicion Tree structure: \n: ", export_text(clf, feature_names=["wickets"]))
```
